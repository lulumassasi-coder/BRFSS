# Data Pre-Processing

```python
# Packages

import os
os.chdir("/Users/kingmushi/Documents/Oswald Mushi/Consultancy/LuLu/ML Project")

import pandas as pd
import numpy as np
import  pyreadstat
```

# 2022

```python
# Data Import
brfss2022 = pd.read_sas("Dataset/BRFSS2022/LLCP2022.XPT")
```

```python
# Selecting relevant variables

col=["_AGE_G", "SEXVAR","_BMI5","_RFBMI5","HTM4", "_TOTINDA","WTKG3", "_RFSMOK3",
        "_EDUCAG" , "_INCOMG1","DIABETE4", "ASTHNOW", "CHCKDNY2", "_DRDXAR2","_RFDRHV8"]

data2022=brfss2022[col]

data2022.head()
```

```python
# Creating the missing variables

data2022["High_Cholesterol"] = float(999)
data2022["High_BP"] = float(999)
```

```python
# Renaming Variables

data2022 = data2022.rename(columns={ "SEXVAR": "Sex",
                           "_BMI5": "BMI",
                           "_AGE_G": "Age_group",
                           "_RFBMI5": "Obese",
                           "HTM4": "Height_in_m",
                           "_TOTINDA": "Physical_Activity",
                          "_RFDRHV8": "Heavy_Drinker",
                           "WTKG3": "Weight_in_kg",
                           "_RFSMOK3": "Smoker",
                           "_EDUCAG": "Education_level",
                           "_INCOMG1": "Income_Cat",
                             "DIABETE4": "Diabetes",
                             "ASTHNOW": "Asthma",
                             "CHCKDNY2": "Kidney_Disease",
                             "_DRDXAR2": "Arthritis"
                         })
data2022.info()
```

```python
# Exporting The file
pyreadstat.write_dta(data2022 ,dst_path = "Output/BRFSS22.dta")
```
